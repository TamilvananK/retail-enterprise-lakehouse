```python
tables_to_load = {
    "dbt_retail_dev.gold.dim_customer": "analytics_gold.dim_customer",
    "dbt_retail_dev.gold.dim_product": "analytics_gold.dim_product",
    "dbt_retail_dev.gold.fact_sales": "analytics_gold.fact_sales",
    "dbt_retail_dev.gold.fact_inventory": "operations_gold.fact_inventory",
    "dbt_retail_dev.gold.inventory_health": "operations_gold.inventory_health"
}
```

```python
for source_table, target_destination in tables_to_load.items():

    df = spark.table(source_table)
    write_options = {
        **{k: v for k, v in snowflakeOptions.items() if k != "sfUrl"},
        "host": snowflakeOptions["sfUrl"]
    }

    df.write \
      .format("snowflake") \
      .options(**write_options) \
      .option("dbtable", target_destination) \
      .mode("overwrite") \
      .save()
```
