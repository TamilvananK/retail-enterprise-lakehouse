```python
dbutils.widgets.text("dataset_name","")
dbutils.widgets.text("source_format","")
dbutils.widgets.text("source_path","")
dbutils.widgets.text("target_path","")
dbutils.widgets.text("load_type","")

dataset_name = dbutils.widgets.get("dataset_name")
source_format = dbutils.widgets.get("source_format")
source_path = dbutils.widgets.get("source_path")
target_path = dbutils.widgets.get("target_path")
load_type = dbutils.widgets.get("load_type")
source_path = f"abfss://retail-project@sourcedataretail.dfs.core.windows.net/{source_path}"
target_path = f"abfss://retail-project@sourcedataretail.dfs.core.windows.net/{target_path}"
```

```python
print(dataset_name)
print(source_format)
print(source_path)
print(target_path)
print(load_type)
```

```python
if source_format == "csv":

    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(source_path)
    )

elif source_format == "json":

    df = spark.read.json(source_path)

elif source_format == "parquet":

    df = spark.read.parquet(source_path)

else:
    raise Exception(f"Unsupported format {source_format}")
```

```python


if load_type == "full":
    df.write.format("delta").mode("overwrite").save(target_path)

elif load_type == "incremental":
    
    (
        df.write
          .format("delta")
          .mode("append")
          .save(target_path)
    )
    # watermark logic
    # MERGE logic
```
